### Problem Statement 1:Smart Parking Lot Management System


```python
def smart_parking(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

        # track peak usage
        if parked > peak:
            peak = parked

    # parking status
    if parked > capacity:
        status = "Overflow Alert!"
    elif parked == capacity:
        status = "Full"
    else:
        status = "Available"

    return parked, peak, status


# Sample Input
capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

current, peak_usage, status = smart_parking(capacity, logs)

print("Currently Parked Vehicles:", current)
print("Peak Usage:", peak_usage)
print("Parking Status:", status)
```

### Problem Statement 2:Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic, weather):
    # base time
    time = distance * 5

    # traffic delay
    if traffic == "Medium":
        time += 5
    elif traffic == "High":
        time += 15

    # weather delay
    if weather == "Rainy":
        time += 10
    elif weather == "Storm":
        time += 20

    return time


# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

eta = estimate_delivery_time(distance, traffic, weather)

print("Estimated Delivery Time:", eta, "minutes")
```

### Problem Statement 3:Movie Theatre Seat Occupancy Analyzer


```python
def analyze_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    # Determine status
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Available"

    # Suggest additional shows
    if occupancy >= 90:
        suggestion = "Consider opening additional shows"
    else:
        suggestion = "No extra shows needed"

    return occupancy, status, suggestion


# 🔹 User Input
total_seats = int(input("Enter total seats: "))
booked_count = int(input("Enter number of booked seats: "))

# Create dummy list to simulate bookings
booked_seats = [1] * booked_count

occupancy, status, suggestion = analyze_occupancy(total_seats, booked_seats)

print(f"Occupancy: {occupancy:.0f}%")
print("Show Status:", status)
print("Suggestion:", suggestion)
```

### Problem Statement 4:Cloud Server Load Classification System


```python
def classify_server_load(cpu_readings):
    avg_cpu = sum(cpu_readings) / len(cpu_readings)

    # classification
    if avg_cpu < 50:
        status = "Normal"
    elif avg_cpu <= 80:
        status = "Warning"
    else:
        status = "Critical"

    return avg_cpu, status


# Sample Input
cpu_readings = [45, 60, 70, 85, 90]

avg, status = classify_server_load(cpu_readings)

print(f"Average CPU Load: {avg:.0f}%")
print("Server Status:", status)
```

### Problem Statement 5:Smart Classroom Resource Usage Monitor


```python
def monitor_resources(usage):
    overused = []

    for resource, hours in usage.items():
        if hours > 8:
            overused.append(resource)

    alert = "Yes" if overused else "No"

    return overused, alert


# Sample Input
usage_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

overused_resources, alert = monitor_resources(usage_data)

print("Overused Resources:", ", ".join(overused_resources))
print("Energy Alert:", alert)
```

###  Problem Statement 6: Online Event Registration Capacity Controller


```python
def manage_registrations(capacity, registrations):
    if registrations <= capacity:
        confirmed = registrations
        waitlist = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"

    return confirmed, waitlist, status


# User Input
capacity = int(input("Enter event capacity: "))
registrations = int(input("Enter number of registrations: "))

confirmed, waitlist, status = manage_registrations(capacity, registrations)

print("Confirmed Registrations:", confirmed)
print("Waitlisted Users:", waitlist)
print("Registration Status:", status)
```

# 1 . User Login Check

```python
username = input("Enter Username: ")
password = input("Enter Password: ")
if username == "admin" and password == "1234":
  print("Login Successful")
else:
  print("Invalid Credentials")
```

# 2. Pass / Fail Analyzer

```python
marks = [45, 78, 90, 33, 60]
pas = 0
fail = 0
for i in marks:
  if i >= 50:
    pas += 1
  else:
    fail += 1
print("Pass: ", pas)
print("Fail: ", fail)

```

# 3 . Simple Data Cleaner


```python
names = [" Alice ", "bob", " CHARLIE "]
name = []
for i in names:
  name.append(i.strip().lower())
print(name)
```

# 4. Message Length Analyzer

```python
messages = ["Hi", "Welcome to the platform", "OK"]
for i in messages:
  if len(i) > 10:
    print(f" '{i}'  is more than 10 characters")
  else:
    print(f" '{i}'  is less than 10 characters")
```

# 5. Error Message Detector


```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]
count = 0
for  i in logs:
  if i == "ERROR":
    count += 1
print(f"Error Detected {count} Times")
```

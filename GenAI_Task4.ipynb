## Problem 1: Mobile Recharge Validation System


```python
def recharge():
  print("This Are The Recharge Plan We Provide:  199, 299, 399, 599 ")
  while True:
    amount = int(input("Enter Recharge Amount: "))
    if 50 <= amount:
      if amount in Valid_plans:
        if amount == amount:
          return "Recharge Successful"
          break
      else:
        print("Invalid Recharge Amount Enter Amount Between Plan")
    else:
      print('Please enter amount greater than 50')

Valid_plans=[199, 299, 399, 599]
recharge()
```

## Problem 2: Inventory Reorder Alert System

```python
def check_inventory(products):
    for item, stock in products.items():
        if stock < 15:
            print(f"{item}:  Reorder Alert")
        else:
            print(f"{item}:  Stock OK")

inventory = {"Rice": 10, "Sugar": 25, "Salt": 12, "Oil": 30}
check_inventory(inventory)
```

## Problem 3: Student Result Processing System

```python
def marks():
  subjects = 6
  mark = []
  a = 1
  while len(mark) < subjects:
    sub = int(input(f"Enter Marks of subject no {a} : "))
    mark.append(sub)
    a += 1
  avg = sum(mark)/subjects
  if avg >= 50:
    return 'Pass'
  else:
    return 'Fail'

marks()
```

## Problem 4: Cab Fare Estimator with Retry Option


```python
def fare():
  base_fare = 50
  extra_charge = 0
  peak_hour = [i for i in range(6,9)]
  while True:
    distance = int(input("Enter Distance in KM: "))
    hour = int(input("Enter Time in Hours: "))
    fare_amount = base_fare + (distance * 12)
    if hour in peak_hour:
      extra_charge = fare_amount * 0.25
      fare_amount = fare_amount + extra_charge
    check = input(f'Final amount will be {fare_amount} do want to Re-enter detail Y/N \n')
    if check == 'N':
      return 'Your Cab has booked succesfully enjoy the ride :) '
    else:
      continue
fare()
```

## Problem 5: Employee Attendance Eligibility Checker


```python
def check_attendance(attendance):
    present = 0

    for day in attendance:
        if day == "P":
            present += 1

    percentage = (present / len(attendance)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

print(check_attendance(["P","P","A","P","P","A","P"]))
```

## Problem 6: Password Strength Checker


```python
def check_password(password):
    special_chars = "@#$"
    has_digit = False
    has_special = False

    if len(password) < 8:
        return "Weak Password"

    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

print(check_password("Pass@123"))
```
